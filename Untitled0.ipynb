```python
import pandas as pd

# Sample civic incident data
data = {
    "Incident_Type": ["Traffic", "Public Service", "Traffic", "Noise", "Public Service", "Traffic"],
    "Response_Time_Min": [12, 5, 18, 8, 4, 15]
}

df = pd.DataFrame(data)

# Calculate average response time by incident type
print(df.groupby("Incident_Type").mean())
```

```python
import sqlite3
import pandas as pd

# 1. Connect to our in-memory SQL database
conn = sqlite3.connect(":memory:")

# 2. Create Table 1: Intelligence Logs (Incidents & Threat Levels)
conn.execute("""
CREATE TABLE intelligence_logs (
    case_id TEXT PRIMARY KEY,
    agency TEXT,
    incident_type TEXT,
    threat_level TEXT,
    response_time_min INT
);
""")

# 3. Create Table 2: Field Agents (Assigned Federal Personnel)
conn.execute("""
CREATE TABLE field_agents (
    agent_id TEXT,
    case_id TEXT,
    agent_name TEXT,
    clearance_level TEXT
);
""")

# 4. Insert Mock Incident Data
mock_incidents = [
    ("CASE-101", "FBI", "Cyber Incident", "High", 12),
    ("CASE-102", "Local PD", "Traffic Stop", "Low", 5),
    ("CASE-103", "FBI", "Financial Fraud", "High", 18),
    ("CASE-104", "CIA", "Counter-Intel", "Critical", 8),
    ("CASE-105", "Local PD", "Public Disturbance", "Low", 4)
]
conn.executemany("INSERT INTO intelligence_logs VALUES (?, ?, ?, ?, ?)", mock_incidents)

# 5. Insert Mock Field Agent Assignments
mock_agents = [
    ("AGT-09", "CASE-101", "Special Agent Miller", "Top Secret"),
    ("AGT-14", "CASE-103", "Special Agent Vance", "Secret"),
    ("AGT-01", "CASE-104", "Officer Chen", "Top Secret / SCI")
]
conn.executemany("INSERT INTO field_agents VALUES (?, ?, ?, ?)", mock_agents)

# 6. SQL CROSS-TABLE JOIN QUERY:
# Connect incidents with assigned agents where threat level is High or Critical
sql_join_query = """
SELECT
    logs.case_id,
    logs.agency,
    logs.incident_type,
    logs.threat_level,
    agents.agent_name,
    agents.clearance_level
FROM intelligence_logs AS logs
LEFT JOIN field_agents AS agents
  ON logs.case_id = agents.case_id
WHERE logs.threat_level IN ('High', 'Critical')
  AND logs.agency IN ('FBI', 'CIA');
"""

# Run the query and view results in pandas
join_results = pd.read_sql_query(sql_join_query, conn)
print("--- HIGH THREAT CASES WITH ASSIGNED AGENTS (SQL JOIN) ---")
print(join_results)
```

```python
# 1. Add Priority Level and SLA (Service Level Agreement) Thresholds
data["Priority"] = ["High", "Low", "High", "Low", "Medium", "High"]
df = pd.DataFrame(data)

# 2. Flag calls as "Delayed" if response time exceeds 10 minutes
df["Status"] = df["Response_Time_Min"].apply(lambda x: "Delayed" if x > 10 else "On Time")

# 3. Filter and view ONLY High-Priority incidents
high_priority_df = df[df["Priority"] == "High"]

print("--- ALL INCIDENTS WITH STATUS ---")
print(df)

print("\n--- HIGH PRIORITY ONLY ---")
print(high_priority_df)
```

```python
import sqlite3
import pandas as pd

# 1. Connect to an in-memory SQL database
conn = sqlite3.connect(":memory:")

# 2. Create a mock FBI/Civic intelligence table
query_create_table = """
CREATE TABLE intelligence_logs (
    case_id TEXT,
    agency TEXT,
    incident_type TEXT,
    threat_level TEXT,
    response_time_min INT
);
"""
conn.execute(query_create_table)

# 3. Insert mock records into SQL
mock_data = [
    ("CASE-101", "FBI", "Cyber Incident", "High", 12),
    ("CASE-102", "Local PD", "Traffic Stop", "Low", 5),
    ("CASE-103", "FBI", "Financial Fraud", "High", 18),
    ("CASE-104", "CIA", "Counter-Intel", "Critical", 8),
    ("CASE-105", "Local PD", "Public Disturbance", "Low", 4)
]
conn.executemany("INSERT INTO intelligence_logs VALUES (?, ?, ?, ?, ?)", mock_data)

# 4. Write a SQL Query: Select all HIGH or CRITICAL threats handled by Federal Agencies
sql_query = """
SELECT case_id, agency, incident_type, threat_level
FROM intelligence_logs
WHERE threat_level IN ('High', 'Critical')
  AND agency IN ('FBI', 'CIA');
"""

# Run the SQL query and view results in pandas!
sql_results = pd.read_sql_query(sql_query, conn)
print("--- HIGH THREAT FEDERAL CASES (SQL QUERY RESULT) ---")
print(sql_results)
```
