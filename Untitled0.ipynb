```python
import pandas as pd

# Sample civic incident data
data = {
    "Incident_Type": ["Traffic", "Public Service", "Traffic", "Noise", "Public Service", "Traffic"],
    "Response_Time_Min": [12, 5, 18, 8, 4, 15]
}

df = pd.DataFrame(data)

# Calculate average response time by incident type
print(df.groupby("Incident_Type").mean())
```

```python
# 1. Add Priority Level and SLA (Service Level Agreement) Thresholds
data["Priority"] = ["High", "Low", "High", "Low", "Medium", "High"]
df = pd.DataFrame(data)

# 2. Flag calls as "Delayed" if response time exceeds 10 minutes
df["Status"] = df["Response_Time_Min"].apply(lambda x: "Delayed" if x > 10 else "On Time")

# 3. Filter and view ONLY High-Priority incidents
high_priority_df = df[df["Priority"] == "High"]

print("--- ALL INCIDENTS WITH STATUS ---")
print(df)

print("\n--- HIGH PRIORITY ONLY ---")
print(high_priority_df)
```

```python
import sqlite3
import pandas as pd

# 1. Connect to an in-memory SQL database
conn = sqlite3.connect(":memory:")

# 2. Create a mock FBI/Civic intelligence table
query_create_table = """
CREATE TABLE intelligence_logs (
    case_id TEXT,
    agency TEXT,
    incident_type TEXT,
    threat_level TEXT,
    response_time_min INT
);
"""
conn.execute(query_create_table)

# 3. Insert mock records into SQL
mock_data = [
    ("CASE-101", "FBI", "Cyber Incident", "High", 12),
    ("CASE-102", "Local PD", "Traffic Stop", "Low", 5),
    ("CASE-103", "FBI", "Financial Fraud", "High", 18),
    ("CASE-104", "CIA", "Counter-Intel", "Critical", 8),
    ("CASE-105", "Local PD", "Public Disturbance", "Low", 4)
]
conn.executemany("INSERT INTO intelligence_logs VALUES (?, ?, ?, ?, ?)", mock_data)

# 4. Write a SQL Query: Select all HIGH or CRITICAL threats handled by Federal Agencies
sql_query = """
SELECT case_id, agency, incident_type, threat_level
FROM intelligence_logs
WHERE threat_level IN ('High', 'Critical')
  AND agency IN ('FBI', 'CIA');
"""

# Run the SQL query and view results in pandas!
sql_results = pd.read_sql_query(sql_query, conn)
print("--- HIGH THREAT FEDERAL CASES (SQL QUERY RESULT) ---")
print(sql_results)
```
