```python
import pandas as pd

# Sample civic incident data
data = {
    "Incident_Type": ["Traffic", "Public Service", "Traffic", "Noise", "Public Service", "Traffic"],
    "Response_Time_Min": [12, 5, 18, 8, 4, 15]
}

df = pd.DataFrame(data)

# Calculate average response time by incident type
print(df.groupby("Incident_Type").mean())
```

```python
# 1. Add Priority Level and SLA (Service Level Agreement) Thresholds
data["Priority"] = ["High", "Low", "High", "Low", "Medium", "High"]
df = pd.DataFrame(data)

# 2. Flag calls as "Delayed" if response time exceeds 10 minutes
df["Status"] = df["Response_Time_Min"].apply(lambda x: "Delayed" if x > 10 else "On Time")

# 3. Filter and view ONLY High-Priority incidents
high_priority_df = df[df["Priority"] == "High"]

print("--- ALL INCIDENTS WITH STATUS ---")
print(df)

print("\n--- HIGH PRIORITY ONLY ---")
print(high_priority_df)
```
